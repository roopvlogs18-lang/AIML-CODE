```python
from collections import deque

# 1. Adjacency List Representation of the Tree Graph
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': [],
    'F': []
}

# 2. Breadth-First Search (BFS) Function
def bfs(graph, start):
    visited = []
    queue = deque([start])

    while queue:
        node = queue.popleft()
        if node not in visited:
            visited.append(node)
            # Add adjacent unvisited nodes to the queue
            for neighbor in graph[node]:
                if neighbor not in visited:
                    queue.append(neighbor)
    return visited

# 3. Depth-First Search (DFS) Function
def dfs(graph, start, visited=None):
    if visited is None:
        visited = []

    if start not in visited:
        visited.append(start)
        # Recursively visit deep into each branch
        for neighbor in graph[start]:
            dfs(graph, neighbor, visited)

    return visited

# Execute the traversals
bfs_result = bfs(graph, 'A')
dfs_result = dfs(graph, 'A')

# 4. Output the results
print("BFS Traversal starting from A:", " -> ".join(bfs_result))
print("DFS Traversal starting from A:", " -> ".join(dfs_result))
```
